**Мета:** отримати базові знання та навички роботи зі списками Python.
**Варіант:** 3

### Завдання 1
Введіть список цілих чисел із n>2 різними елементами. Визначити індекси максимального та мінімального елементів.

### Завдання 2
Знайти суму цілих чисел з парними індексами та помножити на останній елемент списку.

### Завдання 3
Знайти відсутній елемент у невідсортованій арифметичній прогресії.

```python
def task_1():
    n = int(input("Введіть n: "))
    mas = [int(x) for x in input("Введіть числа через пробіл: ").split()]

    in_mx = mas.index(max(mas))
    in_mn = mas.index(min(mas))

    print(f"{in_mx},{in_mn}")

task_1()
```

```python
def task_2():
    ryadok = input("Введіть список (або Enter, якщо порожній): ")
    if not ryadok:
        return 0

    mas = [int(x) for x in ryadok.split()]

    suma_parnyh = sum(mas[::2])
    return suma_parnyh * mas[-1]

print(task_2())
```

```python
def task_3():
    mas = sorted([int(x) for x in input("Введіть список: ").split()])
    krok = (mas[-1] - mas[0]) // len(mas)

    for i in range(len(mas) - 1):
        if mas[i+1] - mas[i] != krok:
            return mas[i] + krok

print(task_3())
```
